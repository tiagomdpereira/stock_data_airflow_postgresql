```python
import json

with open("../data/stock_data.json", "r") as f:
    data = json.load(f)

data
```

```python
import pandas as pd

df = pd.json_normalize(data)
df
```

```python
df.info()
```

```python
df[["c", "t"]]
```

```python
df.rename(columns={
    "c": "current_price",
    "d": "difference",
    "dp": "percent_different",
    "h": "high_price_day",
    "l": "low_price_day",
    "o": "open_price_day",
    "pc": "previous_close_price",
    "t": "timestamp"
}, inplace=True)

df
```

```python
df.timestamp = pd.to_datetime(df.timestamp, unit="s", utc=True).dt.tz_convert("Europe/Lisbon")
df
```
